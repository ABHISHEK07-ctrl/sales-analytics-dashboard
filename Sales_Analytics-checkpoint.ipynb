# 8. Return Analysis

This section analyzes returned and non-returned orders and calculates the
overall order return rate.

```python
return_analysis = (
    df.groupby("returned")
      .agg(
          orders=("order_id", "count"),
          sales=("total_amount", "sum"),
          profit=("profit_margin", "sum")
      )
      .reset_index()
)

return_analysis
```

### Return Rate

The overall order return rate is approximately 5.52%, with 1,903 returned
orders out of 34,500 total orders.

```python
return_rate = (
    (df["returned"] == "Yes").sum()
    / len(df)
) * 100

print(f"Return Rate: {return_rate:.2f}%")
```

```python
plt.figure(figsize=(7, 5))

plt.bar(
    return_analysis["returned"],
    return_analysis["orders"]
)

plt.title("Returned vs Non-Returned Orders")
plt.xlabel("Returned")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

# 11. Customer Analysis

This section identifies the top customers based on their total spending.

```python
top_customers = (
    df.groupby("customer_id")
      .agg(
          total_orders=("order_id", "count"),
          total_spent=("total_amount", "sum")
      )
      .sort_values("total_spent", ascending=False)
      .head(10)
      .reset_index()
)

top_customers
```
